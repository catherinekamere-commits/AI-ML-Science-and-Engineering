Problem Statement

Access to reliable water, sanitation and hygiene (WASH) services is an important component of safe and effective healthcare delivery. However, the availability and quality of WASH services can vary across healthcare facilities.

This project aims to explore whether machine learning can be used to predict the level of water service available at healthcare facilities across Africa. The dataset contains information about healthcare facility characteristics, location, sanitation, hygiene, waste management and cleaning practices.

The machine learning problem will be treated as a **multiclass classification task**, where the model will predict one of three water service levels: **no service, limited service, or basic service**.


 Intended AI/ML Outcome

The intended outcome is to develop a machine learning classification model that can predict the **water service level** of a healthcare facility based on available facility and WASH-related characteristics.

The model will classify facilities into one of three categories:

* **No service**
* **Limited service**
* **Basic service**

The longer-term aim is to explore whether machine learning could support the identification of healthcare facilities that may require greater attention or investment in WASH infrastructure.



Data Requirements

To address the problem, data are required on characteristics that may be associated with WASH service availability in healthcare facilities.

The dataset should include information on:

* Country and geographic location
* Healthcare facility type
* Year of observation
* Sanitation facilities and usability
* Number and accessibility of toilets
* Handwashing facilities
* Availability of soap and alcohol hand rub
* Waste management and disposal
* Cleaning protocols, supplies and staff training
* Water service level, which will be used as the target variable

The selected dataset contains these types of information for healthcare facilities across 14 African countries.


Dataset

The project uses the **Africa Synthetic WASH in Healthcare Facilities** dataset obtained from Hugging Face.

The dataset contains **15,000 healthcare facility records** across 14 African countries.

The dataset includes variables covering:

* Facility characteristics
* Geographic location
* Water access and reliability
* Sanitation
* Hygiene
* Waste management
* Cleaning practices
* WASH service-level indicators

The dataset was imported directly into Google Colab for exploration and analysis.



Initial Data Exploration

Initial exploration was performed to understand the structure, completeness and distribution of the dataset.

The dataset contains **15,000 observations**. The target variable, `water_service_level`, contains three categories:

* **No service:** 6,940 facilities (46.3%)
* **Limited service:** 4,353 facilities (29.0%)
* **Basic service:** 3,707 facilities (24.7%)

The proposed predictor variables contained **no missing values**, indicating complete data for the variables considered during the initial assessment.

The dataset contains both categorical and numerical variables, including country, facility type, location, year and number of toilets, as well as several binary WASH indicators.

* `no_service`

### Data Governance Consideration

During the initial exploration, `water_source_type`, `water_reliability` and `has_water_source` were found to have a direct relationship with `water_service_level`. These variables may therefore introduce **target leakage** if used as predictors. They will require careful consideration and may be excluded from the final model to avoid artificially inflated model performance.
